# String methods

String methods return new strings or other results rather than modifying the original. Common operations include `split()`, `join()`, `replace()`, `strip()` and case conversion.



```python
' / '.join(['a', 'b'])  # 'a / b'
'  Hello  '.strip().lower()  # 'hello'
```

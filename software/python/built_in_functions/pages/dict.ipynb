# dict

Create a mutable dictionary from a mapping, key-value pairs, or keyword arguments.

**Signature:** `dict(**kwargs); dict(mapping, /, **kwargs); dict(iterable, /, **kwargs)`



```python
dict(a=1, b=2)  # {'a': 1, 'b': 2}
```

# Knowledge Distillation & Promotion Workflow

How to refine raw daily notes into durable, permanent knowledge.

---



## 🔄 Distillation Cycle (Weekly / Milestone)



```text
[ Raw Daily Notes ] ─── Filter & Synthesize ───> [ Layer 2: Concepts / Cheat Sheets ]
                                             └──> [ Layer 1: Software Guides ]
```



1. **Review:** Look over recent entries in `active_learning/daily_notes/`.
2. **Identify Permanent Knowledge:**
   - Did you learn a universal principle? → Promote to `concepts/`.
   - Did you learn a handy command or syntax shortcut? → Promote to `cheat_sheets/`.
   - Did you solve a tool-specific bug? → Add to `software/<tool>/06-troubleshooting.md`.
3. **Refine:** Polish the promoted note with clear explanations and minimal working examples.

# breakpoint

Call the configured debugger hook, normally entering the debugger.

**Signature:** `breakpoint(*args, **kws)`

Added in Python 3.7; behavior can be controlled by `PYTHONBREAKPOINT` and `sys.breakpointhook`.



```python
breakpoint()
```

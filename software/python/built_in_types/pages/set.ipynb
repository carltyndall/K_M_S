# set

`set` is a mutable, unordered collection of distinct hashable objects. It supports membership tests and operations such as union, intersection and difference.



```python
{1, 2} | {2, 3}  # {1, 2, 3}
```


[Python documentation](https://docs.python.org/3.15/library/stdtypes.html#set-types-set-frozenset)

# min

Return the smallest item, optionally using a key function.

**Signature:** `min(iterable, *, key=None); min(iterable, *, default, key=None); min(arg1, arg2, /, *args, key=None)`

`default` applies only to the single-iterable form.



```python
min([3, 1, 2])  # 1
```

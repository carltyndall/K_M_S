# Effective Note-Taking Habits

Guidelines for sustainable documentation habits.

---



## 💡 Core Principles

1. **Capture First, Distill Later:** Write fast, informal notes in `daily_notes/`. Do not worry about perfection on Day 1.
2. **Write in Your Own Words:** Translating an idea into your own vocabulary forces comprehension and highlights knowledge gaps.
3. **Include the "Why":** Document the reason behind an architectural choice or fix, not just the code snippet.
4. **Record Errors & Fixes:** Troubleshooting logs are the most valuable reference for your future self.

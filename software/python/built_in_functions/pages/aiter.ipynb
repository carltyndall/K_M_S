# aiter

Return an asynchronous iterator for an asynchronous iterable.

**Signature:** `aiter(async_iterable, /)`

Added in Python 3.10; unlike `iter`, there is no two-argument form.



```python
iterator = aiter(async_iterable)
```


[Python documentation](https://docs.python.org/3/builtins/functions.html#aiter)

# Python keywords: alphabetical index

This reference covers the reserved and soft keywords in the [Python lexical analysis documentation](https://docs.python.org/3/reference/lexical_analysis.html#keywords). Soft keywords are reserved only in specific contexts: `match`, `case` and `_` were introduced in Python 3.10, `type` in 3.12, and `lazy` in 3.15. See also the [grouped index](01-grouped-index.md).

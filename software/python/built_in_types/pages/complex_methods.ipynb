# Complex methods

Complex numbers provide `real` and `imag` components and a `conjugate()` method. `complex.from_number()`, added in Python 3.14, converts a number to complex; ordering comparisons are not defined.



```python
(2 + 3j).conjugate()  # (2-3j)
```

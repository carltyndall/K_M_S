# printf-style string formatting

The `%` operator formats values using conversion specifiers in a string. A tuple supplies multiple positional values; a mapping can supply named values.



```python
'%s: %04d' % ('item', 7)  # 'item: 0007'
```


[Python documentation](https://docs.python.org/3.15/library/stdtypes.html#printf-style-string-formatting)

# 🎓 Courses

Structured learning paths, curricula outlines, and module notes.

---

## 🎯 Purpose

The `courses/` directory in Layer 3 provides dedicated workspaces for structured learning programs, online courses, tutorials, and certifications. It organizes course materials into predictable modules so you can track progression from syllabus to completed exercises.

---

## 🏗️ Standard Course Structure

Every course directory follows the standard 4-file scaffold from [`templates/course_template/`](../../templates/course_template/README.ipynb):

- **`00_overview.ipynb`** — High-level description, prerequisites, and learning outcomes.
- **`01_syllabus.ipynb`** — Module-by-module breakdown, lesson schedule, and progress tracking.
- **`02_notes.ipynb`** — Lecture notes, synthesized key takeaways, and conceptual summaries.
- **`03_exercises.ipynb`** — Hands-on practice tasks, coding exercises, and solution notes.

---

## 🔄 Course Workflow

1. **Scaffold:** Copy [`templates/course_template/`](../../templates/course_template/README.ipynb) into `active_learning/courses/<course_name>/` (named in `snake_case`).
2. **Plan:** Outline modules and objectives in `00_overview.ipynb` and `01_syllabus.ipynb`.
3. **Execute:** Record insights in `02_notes.ipynb` and solve problem sets in `03_exercises.ipynb`.
4. **Distill:** Promote reusable mental models to [`concepts/`](../../concepts/README.ipynb) and tool workflows to [`software/`](../../software/README.ipynb) using the [Distillation Workflow](../../workflows/02_knowledge_distillation_and_promotion_workflow.ipynb).

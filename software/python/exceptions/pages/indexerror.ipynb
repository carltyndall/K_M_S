# IndexError

A [`LookupError`](LookupError.md) raised when a sequence index is out of range; slices are normally truncated instead.



```python
items = ['a']
items[1]  # IndexError
```


See [sequence operations](../../built_in_types/pages/sequence_operations.md). [Python documentation](https://docs.python.org/3/library/exceptions.html#IndexError)

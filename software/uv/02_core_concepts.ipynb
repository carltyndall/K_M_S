# uv Core Concepts

## Architectural Mental Model

uv is an extremely fast Python package and project manager written in Rust, designed to replace `pip`, `pip-tools`, and `virtualenv` workflows with deterministic, high-throughput dependency resolution.

---

## 🏛️ Pure Theory & Decoupled Concepts

> **DRY Architecture Rule:** Broad environment management and isolation theories are centralized in Layer 2 (`concepts/`).

- **Environment Isolation:** [`concepts/environment/01-environment-fundamentals.md`](../../concepts/environment/01_environment_fundamentals.ipynb) — Virtual environment state, path resolution, and interpreter decoupling.
- **Dependency & Configuration Management:** [`concepts/configuration_management/`](../../concepts/configuration_management/README.ipynb) — Declarative dependency specifications and lockfile determinism.

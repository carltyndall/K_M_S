# Numeric types

The main numeric types are arbitrary-precision integers (`int`), floating-point numbers (`float`) and complex numbers (`complex`). Mixed arithmetic widens operands as needed; `//` is floor division and `**` is exponentiation.



```python
7 // 2, 2 ** 3, 1 + 2j  # (3, 8, (1+2j))
```


[Python documentation](https://docs.python.org/3.15/library/stdtypes.html#numeric-types-int-float-complex)

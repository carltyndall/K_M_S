# K_M_S Changelog

All notable changes and milestones for the K_M_S repository are documented in this file.

## [v1.0.0] — 2026-10-24
### Added
- Initial foundational release of **K_M_S** (Teachable Knowledge Management System).
- Implemented the Three-Layer Architecture model (`software/`, Cross-Cutting, `active_learning/`).
- Established 1:1 Markdown (`.md`) and Jupyter Notebook (`.ipynb`) dual-format parity.
- Added starter tools (`software/python/`, `software/git/`).
- Added standard templates (`software_template/`, `concept_template/`, `daily_notes_template/`, `course_template/`).
- Included foundational concepts, cheat sheets, glossary items, and standard workflows.

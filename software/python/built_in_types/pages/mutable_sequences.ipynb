# Mutable sequence types

Mutable sequences support in-place changes such as item or slice assignment, deletion, `append()`, `extend()`, `insert()`, `pop()` and `reverse()`.



```python
items = [1, 2]
items.append(3)  # items is now [1, 2, 3]
```


[Python documentation](https://docs.python.org/3.15/library/stdtypes.html#mutable-sequence-types)

# Boolean operations

`and` and `or` short-circuit and return an operand, not necessarily a Boolean; `not` returns a Boolean. Truth testing follows the object's truth-value rules.



```python
name = '' or 'guest'  # 'guest'
False and (1 / 0)  # False; right side is not evaluated
```

# max

Return the largest item, optionally using a key function.

**Signature:** `max(iterable, *, key=None); max(iterable, *, default, key=None); max(arg1, arg2, /, *args, key=None)`

`default` applies only to the single-iterable form.



```python
max([3, 1, 2])  # 3
```

# 🔗 Active Learning (Layer 3)

The dynamic, hands-on workspace for real-time skill acquisition, active courses, daily logs, and technical experiments.

---

## 🎯 Purpose

While Layer 1 and Layer 2 store permanent, curated knowledge, Layer 3 is your active learning laboratory. It is where you:
- Take raw daily notes and reflect on your progress.
- Track structured course curricula and lessons.
- Run temporary experimental code spikes.
- Manage your backlog of things to learn next.

---

## 📂 Active Learning Subdirectories

- **[`courses/`](./courses/README.ipynb)** — Structured curricula, syllabi, lessons, and exercises.
- **[`daily_notes/`](./daily_notes/README.ipynb)** — Chronological learning journal (`YYYY-MM-DD-daily-notes.md`).
- **[`experiments/`](./experiments/README.ipynb)** — Scratchpad for temporary code spikes and syntax tests.
- **[`learning_queue/`](./learning_queue/README.ipynb)** — Prioritized backlog of topics and articles to study.
- **[`reference_library/`](./reference_library/README.ipynb)** — Curated documentation links, books, and reference guides.
- **[`research/`](./research/README.ipynb)** — Deep-dive exploratory notes and technical scratchpads.

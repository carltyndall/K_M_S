# _

`_` is a soft keyword used as a wildcard in patterns: it matches without binding a name. It remains an ordinary identifier outside patterns. Pattern matching was added in Python 3.10.



```python
match (1, 2):
    case (first, _):
        print(first)  # 1
```


[Python documentation](https://docs.python.org/3/reference/compound_stmts.html#wildcard-patterns)

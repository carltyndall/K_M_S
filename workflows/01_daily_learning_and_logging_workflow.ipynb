# Daily Learning & Logging Workflow

Standard routine for daily study and practice.

---



## 🔄 Daily Routine Steps

1. **Initialize Daily Note:**
   - Create `active_learning/daily_notes/YYYY-MM-DD-daily-notes.md`.
   - Set 1–3 clear goals for the session.
2. **Study & Experiment:**
   - Read/watch lesson material.
   - Run minimal working examples in `active_learning/experiments/`.
3. **Capture Notes & Errors:**
   - Record commands, insights, and error fixes into your daily note.
4. **Commit Daily Snapshot:**
   - Stage and commit your daily note with Git.

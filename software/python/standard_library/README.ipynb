# Python standard library essentials: alphabetical index

These pages introduce six useful standard-library modules with examples, not exhaustive API references. They require no third-party packages; consult the linked documentation for version-specific behavior. See also the [grouped index](01-grouped-index.md).

- [`collections`](pages/collections.md) — specialized containers
- [`datetime`](pages/datetime.md) — dates, times, and durations
- [`itertools`](pages/itertools.md) — tools for composing iterators
- [`json`](pages/json.md) — JSON encoding and decoding
- [`pathlib`](pages/pathlib.md) — filesystem paths
- [`re`](pages/re.md) — regular expressions

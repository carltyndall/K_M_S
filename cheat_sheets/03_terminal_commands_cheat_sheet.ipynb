# Terminal Commands Cheat Sheet

Cross-platform comparison between Windows PowerShell and POSIX (Bash/macOS).

| Action | PowerShell | POSIX (Bash / Zsh) |
| :--- | :--- | :--- |
| Print Working Directory | `pwd` / `Get-Location` | `pwd` |
| List Files | `dir` / `ls` / `Get-ChildItem` | `ls -la` |
| Change Directory | `cd <path>` | `cd <path>` |
| Create Folder | `mkdir <name>` | `mkdir -p <name>` |
| Create File | `New-Item <file>` | `touch <file>` |
| Remove File | `rm <file>` | `rm <file>` |
| View File Contents | `cat <file>` / `Get-Content <file>` | `cat <file>` |
| Clear Screen | `clear` / `cls` | `clear` |

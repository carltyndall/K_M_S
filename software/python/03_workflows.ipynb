# Python: Common Workflows

Repeatable operational workflows for creating, testing, and running Python scripts.

## 🔄 Workflow 1: Running a Script
1. Navigate to the project directory:
   ```powershell
   cd software/python/05-projects
   ```
2. Run the script:
   ```powershell
   python hello_world.py
   ```

## 🔄 Workflow 2: Exploring Interactively via REPL
1. Launch REPL:
   ```powershell
   python
   ```
2. Experiment with expressions:
   ```python
   numbers = [1, 2, 3, 4]
   sum(numbers)
   ```
3. Exit REPL:
   ```python
   exit()
   ```

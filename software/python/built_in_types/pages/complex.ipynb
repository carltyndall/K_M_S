# complex

`complex` represents a number with floating-point real and imaginary parts. Complex values support arithmetic but not ordering comparisons.



```python
z = 1 + 2j
z.real, z.imag  # (1.0, 2.0)
```


[Python documentation](https://docs.python.org/3.15/library/stdtypes.html#numeric-types-int-float-complex)

# Comparisons

Comparisons include ordering and equality, identity (`is`, `is not`), and membership (`in`, `not in`). Chained comparisons evaluate each middle expression at most once and short-circuit.



```python
1 < 2 < 3  # True
'a' in 'cat'  # True
```


See [`is`](../../keywords/pages/is.md), [`in`](../../keywords/pages/in.md) and [type comparisons](../../built_in_types/pages/comparisons.md). [Python documentation](https://docs.python.org/3/reference/expressions.html#comparisons)

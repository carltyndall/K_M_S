# Foundational Programming Terms

Essential terminology every programmer should know.

### **Algorithm**
A well-defined sequence of computational steps that transforms an input into an output.

### **Bytecode**
An intermediate, platform-independent representation of source code produced by a compiler (such as Python's CPython engine) to be executed by a virtual machine.

### **Immutable**
An object whose state or value cannot be modified after it is created (e.g., Python `tuple`, `str`, `int`).

### **Mutable**
An object whose state or internal elements can be modified in place after creation (e.g., Python `list`, `dict`).

### **REPL (Read-Eval-Print Loop)**
An interactive programming environment that takes single user inputs, executes them, and returns the result to the user.

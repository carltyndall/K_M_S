# 🧱 Software-Centric Knowledge (Layer 1)

Layer 1 contains documentation and notes organized around specific software tools, runtimes, editors, and operating systems.

---



## 🎯 Purpose

When working with tools like Python, Git, or an IDE, knowledge often starts as practical questions:
- *How do I install this?*
- *What is the mental model for how it operates?*
- *What commands do I run repeatedly?*
- *How do I fix this specific error message?*

Layer 1 captures all tool-specific information in a predictable, standardized format so you can find solutions quickly.

---



## 📂 Standard 7-File Structure

Every software folder in Layer 1 uses the following standard scaffold:

1. **`00-overview.md`** — High-level purpose, why you use the tool, and its role in your workflow.
2. **`01-installation.md`** — Step-by-step setup guides, environment variables, and verification.
3. **`02-core-concepts.md`** — Foundational mental models unique to the software.
4. **`03-workflows.md`** — Repeatable sequences and day-to-day procedures.
5. **`04-commands.md`** — Shortcuts, CLI flags, and quick reference cards.
6. **`05-projects/`** — Lightweight code artifacts, spikes, or sample projects (no `.venv` or `.idea`).
7. **`06-troubleshooting.md`** — Log of encountered errors, root causes, and verified fixes.

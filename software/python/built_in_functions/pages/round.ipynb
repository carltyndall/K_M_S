# round

Round a number to the given number of decimal digits, or to an integer if omitted.

**Signature:** `round(number, ndigits=None)`

Built-in numeric types use ties-to-even rounding; binary floating-point representation can affect results.



```python
round(2.5)  # 2
```


[Python documentation](https://docs.python.org/3/builtins/functions.html#round)

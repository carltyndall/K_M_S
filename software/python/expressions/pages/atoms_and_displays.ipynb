# Atoms and displays

Atoms include names, literals, parenthesized expressions and displays. Square brackets construct a [list](../../built_in_types/pages/list.md); braces construct a [set](../../built_in_types/pages/set.md) or [dictionary](../../built_in_types/pages/dict.md), with `{}` producing a dictionary. Parentheses group expressions; a comma creates a tuple. String literals may also be formatted or template strings.

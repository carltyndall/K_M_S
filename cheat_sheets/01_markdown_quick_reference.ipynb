# Markdown Quick Reference

Essential syntax elements for drafting clean, readable documentation.



## 📝 Headers
```markdown
# Level 1 Heading (Document Title)
## Level 2 Heading (Major Section)
### Level 3 Heading (Subsection)
```



## ✍️ Text Formatting
- **Bold:** `**bold text**`
- *Italic:* `*italic text*`
- `Inline Code:` `` `inline code` ``
- Strikethrough: `~~strikethrough~~`



## 📋 Lists
```markdown
- Unordered item 1
- Unordered item 2
  - Sub-item

1. Ordered item 1
2. Ordered item 2
```



## 🔗 Links & Images
```markdown
[Link Text](path/to/target.ipynb)
![Alt Text](path/to/image.png)
```



## 📊 Tables
```markdown
| Header 1 | Header 2 |
| :--- | :--- |
| Cell 1 | Cell 2 |
```

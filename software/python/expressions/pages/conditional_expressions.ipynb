# Conditional expressions

`x if condition else y` evaluates the condition first and then only the selected branch.



```python
label = 'positive' if 3 > 0 else 'other'  # 'positive'
```


See [`if`](../../keywords/pages/if.md) and [`else`](../../keywords/pages/else.md). [Python documentation](https://docs.python.org/3/reference/expressions.html#conditional-expressions)

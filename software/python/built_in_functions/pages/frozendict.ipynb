# frozendict

Create an immutable mapping from a mapping, key-value pairs, or keyword arguments.

**Signature:** `frozendict(**kwargs); frozendict(mapping, /, **kwargs); frozendict(iterable, /, **kwargs)`

Added in Python 3.15.



```python
frozendict(a=1)
```

# Common sequence operations

Sequences support membership, indexing, slicing, iteration and length; many also support concatenation and repetition. Negative indices count from the end, and slices exclude the stop index.



```python
[1, 2, 3][-1], 'abc'[1:]  # (3, 'bc')
```


[Python documentation](https://docs.python.org/3.15/library/stdtypes.html#common-sequence-operations)
